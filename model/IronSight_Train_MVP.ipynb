```python
from ultralytics import YOLO
# get the model
model = YOLO("yolov8n.pt")
# start auto training , as the library gives this option
results = model.train(data= f"{dataset.location}/data.yaml",
                      epochs = 50 ,
                      imgsz = 640 ,
                      device = 0
                      )
# 0 means on GPU
```
